# Analiza avtomobilov

V tej projektni nalogi bom analiziral oglase za avtomobile s spletne strani DoberAvto.si.

Zbral sem podatke o imenu in znamki avtomobila, letniku, prevoženih kilometrih, vrsti goriva, tipu menjalnika, moči motorja, ceni in vrsti prodajalca.

V nadaljevanju bom preveril, kako so posamezne lastnosti povezane s ceno avtomobila in kakšni avtomobili se najpogosteje pojavljajo med oglasi.

## Priprava podatkov

Najprej uvozim knjižnici, preberem podatke iz datoteke CSV in stolpce, ki vsebujejo števila, pretvorim v številski tip. Pri pretvarjanju cene in kilometrine se vrednosti, ki niso števila, spremenijo v manjkajoče vrednosti.

```python
import pandas as pd
import matplotlib.pyplot as plt

avti = pd.read_csv("podatki/avti.csv")
pd.options.display.max_rows = 10

avti["Letnik"] = avti["Letnik"].astype("Int64")
avti["Prevoženi kilometri"] = pd.to_numeric(avti["Prevoženi kilometri"], errors="coerce").astype("Int64")
avti["Moč motorja"] = avti["Moč motorja"].astype("Int64")
avti["Cena"] = pd.to_numeric(avti["Cena"], errors="coerce").astype("Int64")

avti.head(10)
```

## Povprečna cena glede na letnik

Najprej me zanima, kako se cena avtomobila spreminja glede na letnik. Pričakujem, da bodo novejši avtomobili v povprečju dražji.

```python
povprecna_cena = avti.groupby("Letnik")["Cena"].mean().sort_index()

povprecna_cena.plot(kind="line", marker="o")
plt.title("Povprečna cena avtomobilov glede na letnik")
plt.xlabel("Letnik")
plt.ylabel("Povprečna cena (€)")
plt.grid()
plt.show()
```

Iz grafa lahko vidimo splošen trend spreminjanja povprečne cene skozi letnike. Pri posameznih letnikih lahko pride do večjih odstopanj, ker je pri nekaterih letnikih manj oglasov.

## Cena glede na prevožene kilometre

Poleg starosti avtomobila na njegovo ceno vpliva tudi število prevoženih kilometrov. Pričakujem, da bodo avtomobili z več prevoženimi kilometri praviloma cenejši.

```python
avti.dropna(subset=["Cena", "Prevoženi kilometri"]).plot(kind="scatter", x="Prevoženi kilometri", y="Cena")
plt.title("Cena glede na prevožene kilometre")
plt.xlabel("Prevoženi kilometri")
plt.ylabel("Cena (€)")
plt.grid()
plt.show()
```

Na grafu lahko preverimo, ali je med kilometrino in ceno opazna povezava. Posamezni avtomobili lahko precej odstopajo, saj na ceno poleg kilometrine vplivajo tudi znamka, letnik, moč motorja in druge lastnosti.

## Povprečna cena glede na znamko

Zanima me tudi, ali obstajajo večje razlike v cenah avtomobilov različnih znamk. Zato bom izračunal povprečno ceno za vsako znamko.

```python
povprecna_cena_znamka = avti.groupby("Znamka")["Cena"].mean().sort_values()

povprecna_cena_znamka.plot(kind="barh", figsize=(10, 8))
plt.title("Povprečna cena glede na znamko avtomobila")
plt.xlabel("Povprečna cena (€)")
plt.ylabel("Znamka")
plt.grid(axis="x")
plt.tight_layout()
plt.show()
```

Graf pokaže razlike v povprečni ceni med znamkami. Pri znamkah z zelo malo oglasi je povprečna cena lahko manj zanesljiva, saj nanjo močno vpliva že en sam avtomobil.

## Najpogostejše znamke avtomobilov

Sedaj me zanima, katere znamke predstavljajo največji delež avtomobilov med zajetimi oglasi. Zaradi preglednosti bom prikazal samo znamke, ki predstavljajo več kot 3 % vseh oglasov.

```python
stevilo_znamk = avti["Znamka"].value_counts()
delezi = stevilo_znamk / stevilo_znamk.sum() * 100
znamke_nad_3 = stevilo_znamk[delezi > 3]

znamke_nad_3.plot(kind="pie", autopct="%1.1f%%", figsize=(8, 8))
plt.title("Znamke z več kot 3 % deležem")
plt.ylabel("")
plt.show()
```

Iz grafa lahko vidimo, katere znamke se med zajetimi oglasi pojavljajo najpogosteje. Ker so prikazane samo znamke z več kot 3 % deležem, manj pogoste znamke na grafu niso prikazane.

## Vrsta goriva

Avtomobili uporabljajo različne vrste goriva. Zanima me, katere vrste goriva se med zajetimi oglasi pojavljajo najpogosteje.

```python
vrste_goriva = avti["Tip goriva"].value_counts()

vrste_goriva.plot(kind="pie", autopct="%1.1f%%", figsize=(8, 8))
plt.title("Delež avtomobilov glede na vrsto goriva")
plt.ylabel("")
plt.show()
```

Tortni graf prikazuje deleže različnih vrst goriva med oglasi. Tako lahko hitro primerjamo, katera vrsta goriva prevladuje in katere se pojavljajo redkeje.

## Vrsta prodajalca

Oglase lahko objavljajo fizične ali pravne osebe. Zanima me, katera skupina predstavlja večji delež zajetih oglasov.

```python
vrste_prodajalcev = avti["Prodajalec"].value_counts()

vrste_prodajalcev.plot(kind="pie", autopct="%1.1f%%", figsize=(7, 7))
plt.title("Delež fizičnih in pravnih prodajalcev")
plt.ylabel("")
plt.show()
```

Graf pokaže razmerje med oglasi fizičnih in pravnih oseb. Iz tega lahko ugotovimo, kdo predstavlja večji del ponudbe na spletni strani.

## Cena glede na moč motorja

Zanima me, ali so avtomobili z močnejšim motorjem praviloma dražji. Primerjal bom moč motorja in ceno posameznega avtomobila.

```python
avti.dropna(subset=["Cena", "Moč motorja"]).plot(kind="scatter", x="Moč motorja", y="Cena")
plt.title("Cena glede na moč motorja")
plt.xlabel("Moč motorja (KM)")
plt.ylabel("Cena (€)")
plt.grid()
plt.show()
```

Na grafu lahko opazujemo povezavo med močjo motorja in ceno. Tudi tukaj lahko posamezni avtomobili močno odstopajo, saj cena ni odvisna samo od moči motorja.

## Povprečna cena glede na tip menjalnika

Zanima me, ali obstaja razlika v povprečni ceni avtomobilov z ročnim in avtomatskim menjalnikom.

```python
cena_menjalnik = avti.groupby("Tip menjalnika")["Cena"].mean()

cena_menjalnik.plot(kind="bar")
plt.title("Povprečna cena glede na tip menjalnika")
plt.xlabel("Tip menjalnika")
plt.ylabel("Povprečna cena (€)")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```

Primerjava pokaže, ali se povprečna cena avtomobilov z avtomatskim menjalnikom razlikuje od povprečne cene avtomobilov z ročnim menjalnikom.

## Povprečna cena glede na vrsto goriva

Za konec bom primerjal še povprečne cene avtomobilov glede na vrsto goriva.

```python
povprecna_cena_gorivo = avti.groupby("Tip goriva")["Cena"].mean().sort_values()

povprecna_cena_gorivo.plot(kind="bar")
plt.title("Povprečna cena glede na vrsto goriva")
plt.xlabel("Vrsta goriva")
plt.ylabel("Povprečna cena (€)")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```

Ta primerjava pokaže, ali se med posameznimi vrstami goriva pojavljajo večje razlike v povprečni ceni avtomobilov.
